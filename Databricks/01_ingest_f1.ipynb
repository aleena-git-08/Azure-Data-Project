```python
# ============================================
# 01 - INGEST F1 DATA
# Raw ADLS → Ingested ADLS
# ============================================

raw_path = "abfss://azure-data@azuredataengine.dfs.core.windows.net/Raw/"
ingested_path = "abfss://azure-data@azuredataengine.dfs.core.windows.net/Ingested/"

# Read raw JSON data
df_raw = (
    spark.read
    .option("multiline", "true")
    .json(raw_path)
)

display(df_raw)
```

```python
# Write the raw data into the Ingested layer

df_raw.write \
    .mode("overwrite") \
    .json(ingested_path)

print("F1 data successfully written to Ingested layer.")
```

```python
# Verify the Ingested layer

df_ingested = (
    spark.read
    .option("multiline", "true")
    .json(ingested_path)
)

display(df_ingested)
```

```python
# Write data to the Ingested layer

df_raw.write \
    .mode("overwrite") \
    .json(ingested_path)

print("F1 data successfully written to Ingested layer.")
```
